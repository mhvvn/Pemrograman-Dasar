```python
ujian = input("Apakah Anda lulus ujian? (Lulus/Tidak) : ")

status = ujian.strip().lower() == "lulus"

if status:
    print("Selamat, Anda dinyatakan lulus.")
else:
    print("Mohon maaf, Anda belum lulus.")
    print("Silakan mengikuti program remedi.")
```

Jajan di Kantin

Rani membeli 3 gorengan seharga Rp 2.500 per buah dan 2 es teh seharga Rp 4.000 per gelas. \
Rani membayar dengan uang Rp 20.000.

Tentukan:

1. Total belanja.
2. Uang kembalian.
3. Apakah uang Rani cukup? (True/False)

# Todo
1. Hitung harga gorengan: jumlah × harga satuan.
2. Hitung harga es teh: jumlah × harga satuan.
3. Jumlahkan keduanya menjadi total.
4. Kembalian = uang dibayar dikurangi total.
5. Bandingkan uang dengan total memakai >=.

```python
#Definiskasn_Variabel

harga_gorengan = int (input ("Masukan Harga Satuan Gorengan:") )
jumlah_gorengan = int(input ("Masukan Jumlah Gorengan yg di beli :"))

harga_esteh = int (input ("Masukan Harga Satuan Es Teh:"))
jumlah_esteh =int( input ("Masukan Jumlah Es Teh yg di beli :"))

uang = int (input ("Masukan Uang yg dibawa :"))
```

```python
#totalBelanja
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh
print ("total belanja:",total)


kembalian = uang - total
print ("uang kembalian:",kembalian)

UangCukup = uang >= total
print ("Uang Cukup?",UangCukup)
```

Durasi Film Maraton

Dina menonton film maraton dengan total durasi 275 menit.

Tentukan:

Berapa jam penuh dan berapa menit sisa (dalam format "... jam ... menit").
Total durasi dalam jam desimal.
Apakah durasi tersebut lebih dari 3 jam dan kurang dari 5 jam? (True/False)

#Todo
1. Jam Total  = total menit dibagi 60 tanpa pecahan (//).
2. Menit sisa = sisa pembagian total menit dengan 60 (%).
3. Jam desimal = total menit dibagi 60 dengan pecahan (/).
4. Cek rentang dengan > dan < yang digabung and.

```python
#durasi_dalam_menit

durasitotal_menit = 275

jamtotal = durasitotal_menit // 60

sisah_menit = durasitotal_menit % 60

jam = durasitotal_menit / 60

rentang_durasi = jam > 180 and jam < 300 

print("=================================================")
print ("durasi akhir ", jamtotal, "jam", sisah_menit, "menit")


print()

print ("========Durasi dalam Menit========")
print ("Durasi total dalam menit : ", durasitotal_menit)
print ("Durasi dalam jam :", jamtotal)
print ("sisah durasi menit :", sisah_menit)

print()
print ("Durasi dalam jam (desimal) :", jam)
print ("Durasi dalam rentang 3jam - 5jam :", rentang_durasi)



```

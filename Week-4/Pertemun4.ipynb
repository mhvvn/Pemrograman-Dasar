Jajan di Kantin

Rani membeli 3 gorengan seharga Rp 2.500 per buah dan 2 es teh seharga Rp 4.000 per gelas. Rani membayar dengan uang Rp 20.000.

Tentukan:


Total belanja.
Uang kembalian.
Apakah uang Rani cukup? (True/False)

# Todo
1. Hitung harga gorengan: jumlah × harga satuan.
2. Hitung harga es teh: jumlah × harga satuan.
3. Jumlahkan keduanya menjadi total.
4. Kembalian = uang dibayar dikurangi total.
5. Bandingkan uang dengan total memakai >=.

```python
#Definiskasn_Variabel
#konversi ke Fungis Input(), Untuk Meminta masukan dari User

harga_gorengan = 2500 
jumlah_gorengan = 3
harga_esteh = 4000
jumlah_esteh = 2
uang = 20000
```

```python
#Operasikan

#total Belanja
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh

#uang Kembalian
kembalian = uang - total

#status Uang Cukup
cukup = uang >= total
```

```python
#Tampilkan_Hasil_Perhitungan

print("Total Belanja     :", total ,"Rupiah")
print("Uang Kembalian    :", kembalian ,"Rupiah")
print("Uang Cukup?       :", cukup)
```

Dengan Fungsi Input()

```python
harga_gorengan = int(input("Harga gorengan: "))
jumlah_gorengan = int(input("Jumlah gorengan: "))
harga_esteh = int(input("Harga es teh: "))
jumlah_esteh = int(input("Jumlah es teh: "))
uang = int(input("Uang dibayar: "))
```

```python
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh
```

```python
print("Total     :", total)
print("Kembalian :", uang - total)
print("Cukup?    :", uang >= total)
```

Durasi Film Maraton

Dina menonton film maraton dengan total durasi 275 menit.

Tentukan:

Berapa jam penuh dan berapa menit sisa (dalam format "... jam ... menit").
Total durasi dalam jam desimal.
Apakah durasi tersebut lebih dari 3 jam dan kurang dari 5 jam? (True/False)

#Todo
1. Jam penuh  = total menit dibagi 60 tanpa pecahan (//).
2. Menit sisa = sisa pembagian total menit dengan 60 (%).
3. Jam desimal = total menit dibagi 60 dengan pecahan (/).
4. Cek rentang dengan > dan < yang digabung and.

```python
#durasi_dalam_menit

total_menit = 150
```

```python
#Operasikan

jam = total_menit // 60

sisa_menit = total_menit % 60

jam_desimal = total_menit / 60

dalam_rentang = total_menit >= 180 and total_menit <= 300
```

```python
#Tampilkan_Hasil_Operasi

print("Durasi       :", jam, "jam", sisa_menit, "menit")
print("Jam desimal  :", round(jam_desimal, 2))
print("3 s.d. 5 jam?:", dalam_rentang)
```

```python
#Menggunakan_Fungsi_Input()

total_menit = int(input("Total durasi (menit): "))

print("Durasi :", total_menit // 60, "jam", total_menit % 60, "menit")
```
